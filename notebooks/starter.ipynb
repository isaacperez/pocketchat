# PocketChat Starter Notebook

Use this notebook to explore `pocketchat` modules and run quick experiments.

```python
from pathlib import Path

from pocketchat.data import CharCorpus
```

## Data Exploration

```python
sample_text = Path('../data/tinyshakespeare.txt')
assert sample_text.exists(), f'Missing file: {sample_text}'

corpus = CharCorpus.from_file(str(sample_text), vocab_size=256)
print('num_tokens:', corpus.data.numel())
print('vocab_size:', corpus.vocab_size)
print('first 20 token ids:', corpus.data[:20].tolist())
print('first 20 decoded:', repr(corpus.decode(corpus.data[:20])))
```

```python
batch = corpus.sample_batch(batch_size=4, seq_len=64, device='cpu')
print('batch shape:', tuple(batch.shape))

for i in range(batch.shape[0]):
    print(f'Batch {i}:')
    print(repr(corpus.decode(batch[i])))
    print()
```
